```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
listings = pd.read_csv('listings.csv')
```

```python
listings
```

```python
sns.countplot(x='neighbourhood_group', data=listings)
plt.show()
```

```python
sns.barplot(x='neighbourhood_group', y='price', data=listings)
plt.show()
```

```python
sns.barplot(x='neighbourhood_group', y='price', data=listings, errorbar=('ci', False))
plt.show()
```

```python
plt.hist(listings['price'])
plt.xlabel('price(in US Dollars)')
plt.show()
```

```python
plt.hist(listings['price'], bins=np.arange(0, 1000, 40))
plt.xlabel('price(in US Dollars)')
plt.show()
```

```python
plt.scatter(x=listings['price'], y=listings['number_of_reviews'])
plt.xlabel('price')
plt.ylabel('number of reviews')
plt.show()
```

```python
plt.scatter(x=listings['price'], y=listings['number_of_reviews'])
plt.xlabel('price')
plt.ylabel('number of reviews')
plt.xlim(0,1100)
plt.show()
```

```python
plt.scatter(x=listings['price'], y=listings['number_of_reviews'], s=5)
plt.xlabel('price')
plt.ylabel('number of reviews')
plt.xlim(0,1100)
plt.show()
```
